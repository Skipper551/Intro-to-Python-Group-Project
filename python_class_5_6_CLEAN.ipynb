### Writing your own class

| Word | Meaning | Example |
|---|---|---|
| **class** | the blueprint | `class Product:` |
| **object** (instance) | one thing built from the blueprint | `mug = Product("mug", 10)` |
| **`__init__`** | runs automatically when you create an object; sets up its data | `self.name = name` |
| **attribute** | data stored in the object | `mug.net_price` |
| **method** | a function inside the class; its first parameter is always `self` | `mug.gross_price()` |
| **`self`** | "this particular object" | inside `gross_price`, `self` is `mug` |
| **`__repr__`** | what `print()` shows | `Product('mug', 10)` |

````python
class Product:
    """A product in our shop."""

    def __init__(self, name, net_price, vat_rate=0.23):     # runs when you write Product(...)
        self.name = name                                     # self = "this particular product"
        self.net_price = net_price
        self.vat_rate = vat_rate

    def gross_price(self):                                   # a method = a function inside the class
        return round(self.net_price * (1 + self.vat_rate), 2)

    def __repr__(self):                                      # what print() shows
        return f"Product('{self.name}', {self.net_price})"
````

````python
mug = Product("mug", 10)
tea = Product("tea", 4, vat_rate=0.08)
print(mug, tea)
print(mug.gross_price(), tea.gross_price())

mug.net_price = 20                         # each object keeps its OWN data
print(mug.gross_price(), tea.gross_price())
````

````python
Product.gross_price(mug)                   # what mug.gross_price() really does: self = mug
````

### Inheritance: a new class built on an existing one
`class DiscountBasket(Basket)` means: *a DiscountBasket **is** a Basket* - it gets all of Basket's methods for free,
and you only write what is **new** or **different**. `super()` calls the parent's version of a method.

👉 In the group project after the break, your group's class inherits from a class called `Analysis` in exactly this way.

````python
class DiscountBasket(Basket):                 # a DiscountBasket IS a Basket
    def __init__(self, owner, discount):
        super().__init__(owner)               # first do everything Basket.__init__ does
        self.discount = discount              # then add what is new

    def total(self):                          # replace ONE method, keep all the others
        return round(super().total() * (1 - self.discount), 2)


vip = DiscountBasket("Vera", discount=0.1)
vip.add("mug", 2, 3.5)
vip.add("tea", 1, 5.5)
print(vip, "| pieces:", vip.n_items())       # n_items() and __repr__ were inherited from Basket
````

# Part 2 - Objects in data & ML libraries: `fit()` and `predict()`

Every model in **scikit-learn** (the ML library we use from next class) is an object of a class, and they all work the same way:

```python
model = LinearRegression()      # 1. create the object (its settings)
model.fit(X, y)                 # 2. learn from data  -> stores what it learned, e.g. model.coef_
model.predict(X_new)            # 3. use what it learned
```
Attributes **learned** during `fit()` end with an underscore: `coef_`, `mean_`... - that's how you know they exist only after fitting.

Let's build the simplest possible model ourselves - then you'll know what is inside those objects.

<div class="alert alert-block alert-success">
⚠️ <b>TASK 3 (together) - our first model class</b>
<br>

Let Copilot write the body of `MeanPredictor`, then run the tests. The first test checks the detail that makes `model.fit(y).predict(X)` work.

</div>

````python
import numpy as np


class MeanPredictor:
    """The simplest possible model: always predicts the average of the training data.

    MeanPredictor() creates a model that has not learned anything yet.

    Methods:
        fit(y):     learns the mean of y and stores it in self.mean_ ; returns self
        predict(X): returns a NumPy array with len(X) copies of self.mean_ ;
                    raises RuntimeError if fit() was not called before
    """
    # let Copilot write the body here: __init__, fit, predict
````

````python
model = MeanPredictor()
assert model.fit([10, 20, 30]) is model, "fit() must return self - otherwise model.fit(y).predict(X) breaks"
assert model.mean_ == 20
assert list(model.predict(["a", "b"])) == [20, 20]
assert len(MeanPredictor().fit([1, 2]).predict(range(5))) == 5
print("All tests passed ✅")
````

````python
try:
    MeanPredictor().predict([1, 2, 3])
    print("❌ should have raised RuntimeError")
except RuntimeError as e:
    print("✅", e)
````

**The real thing.** scikit-learn has exactly this model, called `DummyRegressor`:

````python
try:
    from sklearn.dummy import DummyRegressor
    X = [[0], [0], [0]]                  # scikit-learn always wants inputs X (this model ignores them)
    y = [10, 20, 30]
    real = DummyRegressor(strategy="mean").fit(X, y)
    print("scikit-learn:", real.predict([[0], [0]]), "| ours:", MeanPredictor().fit(y).predict([0, 0]))
except ImportError:
    print("scikit-learn is not installed yet - in the terminal:  pip install scikit-learn")
````

A model that always predicts the average is called a **baseline**. Any real model must be better than it -
otherwise it's useless. 👉 In the project, **Card 6** builds a forecasting class and must beat this baseline.

# Part 3 - Files: reading, writing and combining many files

| Format | Read | Write | Good for |
|---|---|---|---|
| **CSV** | `pd.read_csv(path)` | `df.to_csv(path)` | tables; opens everywhere; **only text** - types (dates!) are lost |
| **Excel** | `pd.read_excel(path)` | `df.to_excel(path)` | sharing with non-programmers (needs `openpyxl`) |
| **JSON** | `json.load(file)` / `pd.read_json(path)` | `json.dump(obj, file)` | nested data from web APIs: dicts inside lists inside dicts |
| **text** | `open(path).read()` | `open(path, "w").write(text)` | anything else |

**`with open(...) as f:`** opens a file and **closes it automatically** at the end of the block.
**`pathlib.Path`** builds paths that work on Windows *and* Mac (no more `\` vs `/` problems).

### Dates in the musicians data
Wikidata writes dates as text like `1980-06-02T00:00:00Z`. We keep **one row per person**, convert the dates and compute the age at death:

````python
people = music.drop_duplicates("instance_ofLabel").copy()          # one row per person
for col in ["date_of_birth", "date_of_death"]:
    # format: %Y = year, %m = month, %d = day; "T" and "Z" are just letters in the text
    # errors="coerce": text that is not a valid date becomes NaT ("Not a Time") instead of crashing
    people[col] = pd.to_datetime(people[col], format="%Y-%m-%dT%H:%M:%SZ", errors="coerce")

people["age_at_death"] = (people["date_of_death"] - people["date_of_birth"]).dt.days / 365.25
people["age_at_death"].describe().round(1)
````
